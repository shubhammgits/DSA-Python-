## Next Permutation

```python
#optimal:

arr = [2,1,5,4,3,0,0]
n = len(arr)
ind=-1

for i in range(n-2,-1,-1):
    if arr[i]<arr[i+1]:
        ind = i
        break

if ind == -1:
    arr.reverse()
    print(arr)
    exit()

for i in range(n-1,ind,-1):
    if arr[i]>arr[ind]:
        arr[i],arr[ind] = arr[ind],arr[i]
        break

left = ind+1
right = n-1

while left<right:
    arr[left],arr[right] = arr[right],arr[left]
    left+=1
    right-=1

print(arr)
```
