## Find the Repeating and Missing number

```python
#brute:

arr = [4,3,6,2,1,1]
repeating = -1
missing = -1


for i in range(1,len(arr)+1):
    count=0
    for j in range(0,len(arr)):
        if arr[j]==i:
            count+=1
            
    if count==2:
        repeating = i

    elif count==0:
        missing = i

    if missing!=-1 and repeating!=-1:
        break

print(repeating)
print(missing)
```

```python
#better: hash array

arr = [4,3,6,2,1,1]
repeating = -1
missing = -1
n = len(arr)
freq= [0]*(n+1)

for nums in arr:
    freq[nums]+=1

print(freq)

for i in range(1,n+1):
    if freq[i]==2:
        repeating = i

    elif freq[i]==0:
        missing = i
    
    if repeating!=-1 and missing!=-1:
        break

print(repeating)
print(missing)
```

```python
#optimal1: using maths

arr = [4,3,6,2,1,1]
n = len(arr)

#S - Sn = x-y
#S2 - S2n = x2-y2

Sn = (n*(n+1))//2
S2n = (n*(n+1)*(2*n+1))//6
S = 0
S2 = 0
for i in range(n):
    S += arr[i]
    S2 += arr[i]*arr[i]

val1 = S - Sn          #x - y
val2 = S2 - S2n        #x2 - y2

val2 = val2//val1       #x + y

x = (val1+val2)//2
y = x - val1

print([x,y])
```
