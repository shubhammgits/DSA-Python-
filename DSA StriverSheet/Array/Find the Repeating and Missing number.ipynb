## Find the Repeating and Missing number

```python
#brute:

arr = [4,3,6,2,1,1]
repeating = -1
missing = -1


for i in range(1,len(arr)+1):
    count=0
    for j in range(0,len(arr)):
        if arr[j]==i:
            count+=1
            
    if count==2:
        repeating = i

    elif count==0:
        missing = i

    if missing!=-1 and repeating!=-1:
        break

print(repeating)
print(missing)
```

```python
#better: hashing

arr = [4,3,6,2,1,1]
repeating = -1
missing = -1
n = len(arr)
freq = {}

for num in arr:
    if num in freq:
        freq[num]+=1
    else:
        freq[num]=1

print(freq)

xor1=0
xor2=0

for i in range(1,n+1):
    xor1=xor1^i

for item in freq:
    xor2=xor2^item

missing = xor1^xor2


for item in freq:
    if freq[item]==2:
        repeating = item

print(missing)
print(repeating)
```
