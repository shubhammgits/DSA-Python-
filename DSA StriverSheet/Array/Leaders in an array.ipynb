## Leaders in an array

```python
#brute:

arr = [10,22,12,3,0,6]
n=len(arr)
ans = []

for i in range(n):
    leader=True
    for j in range(i+1,n):
        if arr[j]>arr[i]:
            leader=False
            break

    if leader == True:
        ans.append(arr[i])

print(ans)
```

```python
#optimal:

arr = [10,22,12,3,0,6]
ans = []
maxi = float('-inf')

for i in range(n-1,-1,-1):
    if arr[i]>maxi:
        maxi=arr[i]
        ans.append(arr[i])

ans.reverse()
print(ans)
```
