## 4 Sum

```python
#brute:

arr = [1,0,-1,0,-2,2]
n = len(arr)
s = set()
ans = []

for i in range(n):
    for j in range(i+1,n):
        for k in range(j+1,n):
            for l in range(k+1,n):
                currSum = arr[i]+arr[j]+arr[k]+arr[l]
                
                if currSum==0:
                    temp = [arr[i],arr[j],arr[k],arr[l]]
                    temp.sort()
                    s.add(tuple(temp))


for lists in s:
    ans.append(list(lists))

print(ans)
```

```python
#better: hashmap

arr = [1,2,-1,-2,2,0,-1]
n = len(arr)
ans = []
s = set()
target = 0

for i in range(n):
    for j in range(i+1,n):
        freq = {}
        for k in range(j+1,n):
            fourth = target - (arr[i]+arr[j]+arr[k])

            if fourth in freq:
                temp = [arr[i],arr[j],arr[k],fourth]
                temp.sort()
                s.add(tuple(temp))
            
            freq[arr[k]] = 1

for quadruplet in s:
    ans.append(list(quadruplet))

print(ans)
```

```python
#optimal:

arr = [1,2,-1,-2,2,0,-1]
n = len(arr)
ans = []
target = 0
arr.sort()

for i in range(n):
    if i>0 and arr[i] == arr[i-1]:
        continue

    for j in range(i+1,n):
        if j>i+1 and arr[j] == arr[j-1]:
            continue

        k = j+1
        l = n-1

        while k<l:
            currSum = arr[i]+arr[j]+arr[k]+arr[l]

            if currSum == target:
                ans.append([arr[i],arr[j],arr[k],arr[l]])
                k+=1
                l-=1

                while k<l and arr[k]==arr[k-1]:
                    k+=1
                while k<l and arr[l]==arr[l+1]:
                    l-=1
            elif currSum<0:
                k+=1
            else:
                l-=1

print(ans)
```
