## Longest Consecutive Sequence

```python
#brute 

def ls(arr,num):
    n=len(arr)
    for i in range(n):
        if arr[i]==num:
            return True
    return False

arr = [102,4,100,1,101,3,2,1,1]
n=len(arr)
longest = 1

for i in range(n):
    x = arr[i]
    count=1

    while ls(arr,x+1) == True:
        x=x+1
        count=count+1
```

```python
#better:

arr = [100,102,100,101,101,4,3,2,3,2,1,1,1,2]
lastSmaller = float('-inf')
countCurr = 0
longest = 1
n=len(arr)
arr.sort()

for i in range(n):
    if arr[i]-1 == lastSmaller:
        countCurr+=1
        lastSmaller = arr[i]

    elif arr[i] != lastSmaller:
        countCurr=1
        lastSmaller=arr[i]

    longest = max(longest,countCurr)

longest
```
