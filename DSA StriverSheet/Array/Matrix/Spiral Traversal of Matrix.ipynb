## Spiral Traversal Of Matrix

```python
#optimal:

arr = [[1,2,3],[4,5,6],[7,8,9]]
n = len(arr)
m = len(arr[0])

left = 0
right = m-1
top = 0
bottom = n-1

ans = []

while top<=bottom and left<=right:
    #step1: left to right (top constant)
    for i in range(left,right+1):
        ans.append(arr[top][i])
    top+=1


    #step2: top to bottom (right constant)
    for i in range(top, bottom+1):
        ans.append(arr[i][right])
    right-=1

    #step3: right to left (bottom constant) + edge case(single row matrix)

    if top<=bottom:
        for i in range(right, left-1,-1):
            ans.append(arr[bottom][i])
        bottom-=1


    #step4: bottom to top (left constant) + edge case(innermost spiral will not go from bottom to top)

    if left<=right:
        for i in range(bottom, top-1,-1):
            ans.append(arr[i][left])
        left+=1

print(ans)
```

```python
def product(num):
    prodN = 1
    while num>0:
        prodN = prodN * (num%10)
        num = num//10

    return prodN
```

```python
n = 13
t=4

while product(n)%t != 0:
    n=n+1

    if product(n)%t == 0:
        print(n)
```
