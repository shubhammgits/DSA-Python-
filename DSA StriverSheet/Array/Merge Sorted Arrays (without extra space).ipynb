## Merge Sorted Arrays without extra spaces


```python
#brute: using an extra array

arr1 = [1,3,5,7]
arr2 = [0,2,6,8,9]

n = len(arr1)
m = len(arr2)

temp = []

for i in range(n):
    temp.append(arr1[i])
for i in range(m):
    temp.append(arr2[i])

temp.sort()

for i in range(n):
    arr1[i] = temp[i]
for i in range(m):
    arr2[i] = temp[n+i]

print(arr1)
print(arr2)
```

```python
#brute: strivers
arr1 = [1,3,5,7]
arr2 = [0,2,6,8,9]
n = len(arr1)
m = len(arr2)

arr3 = [0]*(n+m)
left = 0
right = 0
index = 0

while left<n and right<m:
    if arr1[left]<=arr2[right]:
        arr3[index] = arr1[left]
        left+=1
        index+=1

    else:
        arr3[index] = arr2[right]
        right+=1
        index+=1

while left<n:
    arr3[index] = arr1[left]
    index+=1
    left+=1

while right<m:
    arr3[index] = arr2[right]
    right+=1
    index+=1

for i in range(n+m):
    if i<n: 
        arr1[i] = arr3[i]
    else:
        arr2[i-n] = arr3[i]


print(arr1)
print(arr2)

```
