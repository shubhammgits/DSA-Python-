## Pascals Triangle

#### Type 1

```python
def func(n,r):
    res = 1
    for i in range(r):
        res = res * (n-i)
        res = res//(i+1)

    return res
```

```python
#to find the element at row=5, column=3
func(4,2)
```

### Type 2

```python
def func(n,r):
    res=1
    for i in range(r):
        res = res*(n-i)
        res = res//(i+1)

    return res 
```

```python
func(5,2)
```

```python
#brute:

#to print n5h row elements:
n=5

for i in range(n+1):
    print(func(n,i),end=' ')
```

```python
#optimal:

row = 6       #to print 6th row elements
ans = 1
print(ans,end=' ')

for i in range(1,row):
    ans = ans*(row-i)
    ans= ans//i
    print(ans,end=' ')
```
