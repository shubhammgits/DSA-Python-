## Set Matrix Zeroes

```python
#brute:

arr = [[1,1,1],[1,0,1],[1,1,1]]
n = 3
m = 3

def markRow(i):
    for j in range(m):
        if arr[i][j] != 0:
            arr[i][j] = -1

def markCol(j):
    for i in range(n):
        if arr[i][j]!=0:
            arr[i][j]=-1

for i in range(n):
    for j in range(m):
        if arr[i][j] == 0:
            markRow(i)
            markCol(j)

for i in range(n):
    for j in range(m):
        if arr[i][j] == -1:
            arr[i][j] = 0

print(arr)
```
