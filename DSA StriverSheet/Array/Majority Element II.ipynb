## Majority Element II

```python
#brute:

arr = [2,2,3,3,1,2,2]
n=6

for i in range(n):
    count=0
    for j in range(n):
        if arr[i] == arr[j]:
            count+=1
    
    if count>n//2:
        print(arr[i])
        break
```

```python
#better: using hashmap:

arr = [2,2,3,3,1,2,2]
n = 7

freq = {}
for num in arr:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

for item in freq:
    if freq[item] > n//2:
        print(item)
        break
```
