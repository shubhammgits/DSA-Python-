## Majority Element II

```python
#brute:

arr = [1,1,1,3,3,2,2,2]
n = len(arr)

ans = []

for i in range(n):
    
    if arr[i] not in ans:
        count=0
        for j in range(n):
            if arr[i] == arr[j]:
                count+=1

        if count>n//3:
            ans.append(arr[i])

    if len(ans) == 2:
        break

print(ans)
```
