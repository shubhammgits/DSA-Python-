## Majority Element II

```python
#brute:

arr = [1,1,1,3,3,2,2,2]
n = len(arr)

ans = []

for i in range(n):
    
    if arr[i] not in ans:
        count=0
        for j in range(n):
            if arr[i] == arr[j]:
                count+=1

        if count>n//3:
            ans.append(arr[i])

    if len(ans) == 2:
        break

print(ans)
```

```python
#better: hashmap

arr = [1,1,1,3,3,2,2,2]
n = len(arr)
freq = {}

for num in arr:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

ans = []
for item in freq:
    if freq[item] > n//3:
        ans.append(item)

print(ans)
```

```python
#optimal:

arr = [1,1,1,3,3,2,2,2]
n = len(arr)
elem1 = None
elem2 = None
ans = []

count1=0
count2=0

for i in range(n):
    if count1==0 and arr[i]!=elem2:
        elem1 = arr[i]
        count1 = 1
    
    elif count2==0 and arr[i]!=elem1:
        elem2 = arr[i]
        count2 = 1

    elif elem1 == arr[i]:
        count1+=1
    
    elif elem2 == arr[i]:
        count2+=1

    else:
        count1-=1
        count2-=1


#verification:

verifyCount1 = 0
verifyCount2 = 0

for i in range(n):
    if arr[i] == elem1:
        verifyCount1+=1
    
    elif arr[i] == elem2:
        verifyCount2+=1

if verifyCount1>n//3:
    ans.append(elem1)

if verifyCount2>n//3:
    ans.append(elem2)

print(ans)
```
