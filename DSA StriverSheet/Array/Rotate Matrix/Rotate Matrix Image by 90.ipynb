## Rotate Matrix/Image by 90 degrees

```python
#brute:

arr = [[1,2,3],[4,5,6],[7,8,9]]
n = len(arr)
m = len(arr[0])

ans = [[0 for j in range(n)] for i in range(n)]

for i in range(n):
    for j in range(m):
        ans[j][n-i-1] = arr[i][j]

ans
```

```python
#better:

arr = [[1,2,3],[4,5,6],[7,8,9]]
n = len(arr)
m = len(arr[0])

#transpose:
for i in range(n):
    for j in range(i+1,n):
        arr[i][j], arr[j][i] = arr[j][i], arr[i][j]

#reverse each row:
for i in range(n):
    arr[i].reverse()

arr
```
