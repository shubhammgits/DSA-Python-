## Find out how many times array has been rotated

```python
arr = [4,5,6,7,0,1,2]
n = len(arr)

def func(arr,n):
    low=0
    high=n-1
    ans = float('inf')
    index = -1

    while low<=high:
        mid = low+(high-low)//2

        if arr[low]<=arr[high]:
            if arr[low]<ans:
                index = low
            return index

        if arr[low]<=arr[mid]:
            if arr[low]<ans:
                index = low
            low=mid+1

        else:
            if arr[mid]<ans:
                index =mid
            high=mid-1

    return index


func(arr,n)
```
