## Minimum in a rotate sorted array

```python
# brute: linear search

#optimal: binary search

arr = [4,5,6,7,0,1,2]
n = len(arr)

def func(arr,n):
    low = 0
    high = n-1
    ans = float('inf')

    while low<=high:
        mid = low+(high-low)//2

        if arr[low]<=arr[mid]:
            ans = min(arr[low],ans)
            low=mid+1
        
        elif arr[mid]<=arr[high]:
            ans = min(ans,arr[mid])
            high=mid-1
```
