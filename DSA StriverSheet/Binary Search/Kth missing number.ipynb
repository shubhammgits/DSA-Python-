## Find kth missing number from array

```python
#brute:

arr = [2,3,5,7,11]
k = 5
n = len(arr)

for i in range(n):
    if arr[i]<=k:
        k+=1
    else:
        break

print(k)
```

```python
#binary search:

def func(arr,k):
    n=len(arr)
    low=0
    high=n-1

    while low<=high:
        mid=low+(high-low)//2

        missing = arr[mid] - (mid+1)

        if missing<k:
            low=mid+1
        else:
            high=mid-1

    return high+k+1

arr = [2,3,5,7,11]
k = 5
func(arr,k)
```
