## Painters Partition Problem

```python
#binary search

def check(arr,maxT):
    currT=0
    p=1

    for i in range(len(arr)):
        if currT + arr[i]<=maxT:
            currT+=arr[i]
        else:
            p+=1
            currT=arr[i]

    return p


def bs(arr,k):

    if k>len(arr):
        return -1

    low=max(arr)
    high=sum(arr)

    while low<=high:
        mid=low+(high-low)//2

        pcnt = check(arr,mid)

        if pcnt>k:
            low=mid+1
        else:
            high=mid-1

    return low

arr = [10,20,30,40]
k=2

bs(arr,k)
```
