## Search in a rotated sorted array II

```python
#brute: linear search

#optimal: binary search

arr = [2,5,6,0,0,1,2]
x = 0
n = len(arr)

def func(arr,n,x):
    low = 0
    high = n-1
    while low<=high:
        mid = low+(high-low)//2

        if arr[mid]==x:
            return True

        if arr[low]==arr[mid] and arr[mid]==arr[high]:
            low+=1
            high-=1
            continue

        #left sorted:
        if arr[low]<=arr[mid]:
            if arr[low]<=x<=arr[mid]:
                high=mid-1
            else:
                low=mid+1

        #right sorted:
        else:
            if arr[mid]<=x<=arr[high]:
                low=mid+1
            else:
                high=mid-1

    return False

func(arr,n,x)
```
