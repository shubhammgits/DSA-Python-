## Find Peak Element

```python
#brute:

arr = [1,2,3,4,5,6,7,8,5,1]
n = len(arr)

def func(arr,n):
    for i in range(n):

        if i==0:
            if arr[i]>arr[i+1]:
                return arr[i]

        elif i==n-1:
            if arr[i]>arr[i-1]:
                return arr[i]

        else:
            if arr[i-1]<arr[i] and arr[i]>arr[i+1]:
                return arr[i]

func(arr,n)
```
