## Find Peak Element

```python
#brute:

arr = [1,2,3,4,5,6,7,8,5,1]
n = len(arr)

def func(arr,n):
    for i in range(n):

        if i==0:
            if arr[i]>arr[i+1]:
                return arr[i]

        elif i==n-1:
            if arr[i]>arr[i-1]:
                return arr[i]

        else:
            if arr[i-1]<arr[i] and arr[i]>arr[i+1]:
                return arr[i]

func(arr,n)
```

```python
#optimal: binary search

arr = [1,2,3,4,2,1,5,3,2,1]
n = len(arr)

def func(arr,n):
    if n==1:
        return 0
    if arr[0]>arr[1]:
        return 0
    if arr[n-1]>arr[n-2]:
        return n-1

    low = 1
    high = n-2

    while low<=high:
        mid = low+(high-low)//2

        if arr[mid]>arr[mid-1] and arr[mid]>arr[mid+1]:
            return mid

        elif arr[mid]>arr[mid-1]:
            low=mid+1
        
        elif arr[mid]>arr[mid+1]:
            high=mid-1

        else:
            low=mid+1       #or high=mid-1

    return -1

func(arr,n)
```
