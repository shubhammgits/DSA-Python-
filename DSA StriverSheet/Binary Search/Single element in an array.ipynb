## Single element in a sorted array

```python
#brute:

arr = [1,1,2,2,3,4,4,5,5]
n = len(arr)

def func(arr,n):

    if n==1:
        return arr[0]
    
    for i in range(n):

        if i==0:
            if arr[i]!=arr[i+1]:
                return arr[0]

        elif i==n-1:
            if arr[i]!=arr[i-1]:
                return arr[n-1]

        else:
            if arr[i]!=arr[i-1] and arr[i]!=arr[i+1]:
                return arr[i]

func(arr,n)
```

```python
#better: binary search

arr = [1,1,2,2,3,3,4,5,5,6,6]
n = len(arr)

def func(arr,n):
    if n==1:
        return arr[0]

    if arr[0]!=arr[1]:
        return arr[0]

    if arr[n-1]!=arr[n-2]:
        return arr[n-1]

    low = 1
    high = n-2

    while low<=high:
        mid = low+(high-low)//2

        if arr[mid]!=arr[mid-1] and arr[mid]!=arr[mid+1]:
            return arr[mid]


        if (mid%2==1 and arr[mid]==arr[mid-1]) or (mid%2==0 and arr[mid]==arr[mid+1]):
            low = mid+1

        else:
            high = mid-1

    return -1

func(arr,n)
```
