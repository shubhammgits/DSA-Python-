## Minimum days to make 'm' bouqets

```python
# Brute Force

arr = [7,7,7,7,13,11,12,7]
m = 2
k = 3

def possible(arr, day, m, k):
    cnt = 0
    noOfB = 0

    for i in range(len(arr)):
        if arr[i] <= day:
            cnt += 1
        else:
            noOfB += cnt // k
            cnt = 0

    # For the last consecutive segment
    noOfB += cnt // k

    return noOfB >= m


def minDays(arr, m, k):
    n = len(arr)

    if n < m * k:
        return -1

    mini = min(arr)
    maxi = max(arr)

    for day in range(mini, maxi + 1):
        if possible(arr, day, m, k):
            return day

    return -1


print(minDays(arr, m, k))
```
