```python
arr = [3,7,8,9,10,12,19,23]
target = 12

def bs(arr,target):
    n = len(arr)
    low=0
    high=n-1

    while low<=high:
        mid=low+(high-low)//2

        if arr[mid]==target:
            return mid

        elif arr[mid]>target:
            high=mid-1

        else:
            low=mid+1

bs(arr,target)
```
