```python
arr = [3,7,8,9,10,12,19,23]
target = 12

def bs(arr,target):
    n = len(arr)
    low=0
    high=n-1

    while low<=high:
        mid=low+(high-low)//2

        if arr[mid]==target:
            return mid

        elif arr[mid]>target:
            high=mid-1

        else:
            low=mid+1

bs(arr,target)
```

```python
#lower bound:

arr = [3,5,8,15,15,15,19]
x=15

def lb(arr,x):
    n=len(arr)
    low=0
    high=n-1

    ans=n

    while low<=high:
        mid=low+(high-low)//2

        if arr[mid]>=x:
            high=mid-1
            ans=mid

        else:
            low=mid+1

    return ans

lb(arr,x)
```

```python
#upper bound:

arr = [3,6,7,7,9,9,9,13,15,19,21]
x = 8

def ub(arr,x):
    n=len(arr)
    low=0
    high=n-1
    ans=n

    while low<=high:
        mid=low+(high-low)//2

        if arr[mid]>x:
            high=mid-1
            ans=mid

        else:
            low=mid+1

    return ans

ub(arr,x)
```
