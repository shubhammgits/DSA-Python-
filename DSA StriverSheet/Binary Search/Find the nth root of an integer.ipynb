## Find the nth root of an integer

```python
#linear iteration

def power(i,n):
    ans=1
    for j in range(n):
        ans=ans*i
    return ans

def func(n,m):
    for i in range(m):
        if power(i,n)==m:
            return i
        elif power(i,n)>m:
            break
    return -1

func(3,27)
```

```python
#binary search way:

def power(mid,n,m):
    ans=1
    for i in range(1,n+1):
        ans=ans*mid
        
        if ans>m:
            return 2
    
    if ans==m:
        return 1
    else:            #ans<m
        return 0

def nthRoot(n,m):
    low=1
    high=m

    while low<=high:
        mid=low+(high-low)//2

        midN = power(mid,n,m)

        if midN==1:
            return mid
        
        elif midN==0:
            low=mid+1
        
        else:
            high=mid-1

    return -1

nthRoot(3,27)
```
