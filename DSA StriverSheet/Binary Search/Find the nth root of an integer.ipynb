## Find the nth root of an integer

```python
#linear iteration

def power(i,n):
    ans=1
    for j in range(n):
        ans=ans*i
    return ans

def func(n,m):
    for i in range(m):
        if power(i,n)==m:
            return i
        elif power(i,n)>m:
            break
    return -1

func(3,27)
```

```python
#binary search way:

def power(i,n):
    ans=1
    for j in range(n):
        ans=ans*i
    return ans


def bs(n,m):
    low=1
    high=m

    while low<=high:
        mid=low+(high-low)//2 

        if power(mid,n)==m:
            return mid

        elif power(mid,n)>m:
            high=mid-1
        
        else:
            low=mid+1

    return -1

bs(3,27)
```
