## Split array largest sum

```python
#binary search

def check(arr,maxSum):
    subarrays=1
    currSum=0

    for i in range(len(arr)):
        if currSum + arr[i]<=maxSum:
            currSum+=arr[i]
        else:
            subarrays+=1
            currSum=arr[i]

    return subarrays

def sol(arr,k):
    low=max(arr)
    high=sum(arr)

    while low<=high:
        mid=low+(high-low)//2

        subarrayCnt = check(arr,mid)

        if subarrayCnt>k:
            low=mid+1
        else:
            high=mid-1
    
    return low

arr = [10,20,30,40]
k = 2
sol(arr,k)
```
