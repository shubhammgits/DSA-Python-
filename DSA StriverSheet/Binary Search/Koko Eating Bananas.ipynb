## Koko Eating Bananas

```python
#brute:
import math

piles = [3,6,7,11]
h = 8

def ans(piles,h):
    for i in range(1,max(piles)+1):
        reqTime = func(piles,i)

        if reqTime<=h:
            return i


def func(arr,hour):
    totalHours = 0
    for i in range(0,len(arr)):
        totalHours = totalHours + math.ceil(arr[i]/hour)

    return totalHours



ans(piles,h)
```
