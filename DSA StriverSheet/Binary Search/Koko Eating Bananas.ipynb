## Koko Eating Bananas

```python
#brute:
import math

piles = [3,6,7,11]
h = 8

def ans(piles,h):
    for i in range(1,max(piles)+1):
        reqTime = func(piles,i)

        if reqTime<=h:
            return i


def func(arr,hour):
    totalHours = 0
    for i in range(0,len(arr)):
        totalHours = totalHours + math.ceil(arr[i]/hour)

    return totalHours



ans(piles,h)
```

```python
#binary search:

def bs(piles,h):
    low=1
    high=max(piles)

    ans = float('inf')

    while low<=high:
        mid=low+(high-low)//2

        midHour = func(piles,mid)

        if midHour<=h:
            ans = min(ans,mid)
            high = mid-1
        
        else:
            low=mid+1

    return ans


def func(piles,mid):
    totalHours = 0
    for i in range(0,len(piles)):
        totalHours+=math.ceil(piles[i]/mid)

    return totalHours

    
bs(piles,h)
```
