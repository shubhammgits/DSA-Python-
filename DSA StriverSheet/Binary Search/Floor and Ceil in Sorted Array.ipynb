## Floor and Ceil in Sorted Array

```python
#lower bound = ceil

arr = [10,20,30,40,50]
n = len(arr)
x = 25

def ceil(arr,n,x):
    low = 0
    high = n-1
    ans = n

    while low<=high:
        mid = low+(high-low)//2

        if arr[mid]>=x:
            ans = arr[mid]
            high=mid-1
        else:
            low=mid+1

    return ans

def floor(arr,n,x):
    low = 0
    high = n-1
    ans = n

    while low<=high:
        mid = low+(high-low)//2

        if arr[mid]<=x:
            ans = arr[mid]
            low=mid+1
        else:
            high=mid-1

    return ans

print(ceil(arr,n,x))
print(floor(arr,n,x))
```
