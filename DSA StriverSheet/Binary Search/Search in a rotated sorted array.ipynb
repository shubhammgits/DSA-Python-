## Search in a Rotated Sorted Array

```python
#brute: linear search
```

```python
#better: binary search 

def func(arr,n,x):
    low = 0
    high = n-1
    while low<=high:

        mid = low+(high-low)//2

        if arr[mid] == x:
            return mid

        #identify sorted half

        #left half sorted?
        if arr[low]<=arr[mid]:
            if arr[low]<= x <= arr[mid]:
                high = mid-1
            else:
                low=mid+1

        #right half is sorted
        else:
            if arr[mid]<= x <= arr[high]:
                low = mid+1
            else:
                high = mid-1

    return -1


arr = [1,3,5,7,8]
x = 3
n = len(arr)

func(arr,n,x)
```
