## Aggressive Cows

```python
#linear search:

arr = [0,3,4,7,9,10]
cows=4

def ls(arr,cows):
    arr.sort()

    for dist in range(1,max(arr)-min(arr)+1):

        if canWePlace(arr,dist,cows)==True:
            continue
        else:
            return dist-1


def canWePlace(arr,dist,cows):
    cntCows=1
    lastCow=arr[0]

    for i in range(1,len(arr)):
        if arr[i]-lastCow>=dist:
            cntCows+=1
            lastCow=arr[i]
            
        if cntCows>=cows:
            return True
    
    return False


ls(arr,cows)
```

```python
#binary search:

arr = [0,3,4,7,9,10]
cows=4

def bs(arr,cows):
    arr.sort()
    n=len(arr)
    low=0
    high=arr[-1] - arr[0]

    while low<=high:
        mid=low+(high-low)//2

        if canWePlace(arr,mid,cows) == True:
            low=mid+1
        else:
            high=mid-1

    return high

def canWePlace(arr,dist,cows):
    cntCows=1
    lastCow=arr[0]

    for i in range(1,len(arr)):
        if arr[i]-lastCow>=dist:
            cntCows+=1
            lastCow=arr[i]
        
        if cntCows>=cows:
            return True
    
    return False


bs(arr,cows)
```
