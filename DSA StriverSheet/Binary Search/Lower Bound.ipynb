## Implement Lower Bound and Higher Bound 

```python
arr = [1,2,3,3,5,8,8,10,10,11]
x = 9
n = len(arr)

def bs(arr,target,n):
    low = 0
    high = n-1
    ans = n

    while low<=high:
        mid = low+(high-low)//2

        if arr[mid]>=target:
            ans = mid
            high=mid-1

        else:                           # arr[mid]<ans:
            low=mid+1

    return ans

bs(arr,x,n)

        
```
