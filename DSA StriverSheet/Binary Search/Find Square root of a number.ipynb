## Find Square root of a number

```python
#linear search way:

n=28

for i in range(1,n+1):
    if i*i<=n:
        ans = i
    else:
        break

print(ans)
```

```python
#binary search

n = 28

def func(n):
    low=1
    high=n
    ans=1

    while low<=high:
        mid=low+(high-low)//2

        if mid*mid<=n:
            low=mid+1
            ans=mid
        else:
            high=mid-1

    return ans

func(n)
```
