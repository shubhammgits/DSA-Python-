## Allocate Books

```python
#brute:

def sol(arr,m):

    if m>len(arr):
        return -1
        
    low=max(arr)
    high=sum(arr)

    for pages in range(low,high+1):

        studCnt = check(arr,pages)

        if studCnt<=m:
            return pages
    
    return -1

def check(arr,pages):
    stud=1
    pagesStud=0

    for i in range(len(arr)):

        if pagesStud + arr[i]<=pages:
            pagesStud+=arr[i]

        else:
            pagesStud=arr[i]
            stud+=1
    
    return stud

arr = [25,46,28,49,24]
m=4

sol(arr,m)
```

```python
#binary search


def check(arr,midPage):
    stud=1
    studPages=0

    for i in range(len(arr)):
        if studPages+arr[i]<=midPage:
            studPages+=arr[i]
        else:
            stud+=1
            studPages=arr[i]

    return stud

def bs(arr,m):

    if m>len(arr):
        return -1
    
    low=max(arr)
    high=sum(arr)

    while low<=high:
        midPage=low+(high-low)//2

        midStud = check(arr,midPage)

        if midStud>m:
            low=midPage+1

        else:
            high=midPage-1

    return low

arr = [25,46,28,49,24]
m=4
bs(arr,m)
```
