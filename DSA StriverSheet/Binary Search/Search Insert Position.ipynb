## Search Insert Position

```python
#using lower bound bs:

arr = [1,2,4,7]
x = 2
n = len(arr)

def bs(arr,n,x):
    low = 0
    high = n-1
    ans = n

    while low<=high:
        mid = low+(high-low)//2

        if arr[mid]>=x:
            ans = mid
            high=mid-1
        else:
            low=mid+1
    return ans

bs(arr,n,x)
```
