## Median of two sorted arrays of different sizes

```python
#brute

def sol(arr1,arr2):
    arr3 = []
    i=0
    j=0
    n1=len(arr1)
    n2=len(arr2)

    while i<n1 and j<n2:
        if arr1[i]<arr2[j]:
            arr3.append(arr1[i])
            i+=1
        else:
            arr3.append(arr2[j])
            j+=1

    while i<n1:
        arr3.append(arr1[i])
        i+=1
    while j<n2:
        arr3.append(arr2[j])
        j+=1


    n3 = len(arr3)

    if n3%2==1:
        return arr3[n3//2]
    else:
        return (arr3[n3//2] + arr3[n3//2 - 1])/2

arr1 = [1,3,4,7,10,12]
arr2 = [2,3,6,15]

sol(arr1,arr2)
```

```python
#better:

def sol(arr1,arr2):
    i=0
    j=0
    n1=len(arr1)
    n2=len(arr2)

    n = n1+n2

    ind2 = n//2
    ind1 = ind2-1

    cnt = 0

    elem1 = -1
    elem2 = -1

    while i<n1 and j<n2:
        if arr1[i]<arr2[j]:
            if cnt==ind1:
                elem1 = arr1[i]

            if cnt == ind2:
                elem2 = arr1[i]

            cnt+=1
            i+=1

        else:
            if cnt == ind2:
                elem2 = arr2[j]
            if cnt == ind1:
                elem1 = arr2[j]

            cnt+=1
            j+=1

    while i<n1:
        if cnt==ind1:
            elem1 = arr1[i]
        if cnt==ind2:
            elem2 = arr1[i]

        cnt+=1
        i+=1


    while j<n2:
        if cnt==ind1:
            elem1 = arr2[j]
        if cnt == ind2:
            elem2 = arr2[j]

        cnt+=1
        j+=1

    if n%2==1:
        return elem2
    else:
        return (elem1+elem2)/2


arr1 = [1,3,4,7,10,12]
arr2 = [2,3,6,15]

sol(arr1,arr2)
```
