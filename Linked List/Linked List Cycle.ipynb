## Linked List Cycle

```python
#brute:

def sol(head):
    temp=head
    myset = set()

    while temp is not None:
        if temp in myset:
            return True
        myset.add(temp)
        temp = temp.next

    return False
```

```python
#optimal:

def sol(head):
    slow=head
    fast=head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

        if slow==fast:
            return True

    return False
```
