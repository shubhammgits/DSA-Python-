## Rotate LinkedList

```python
#solution:

def sol(head,k):

    if head is None or head.next is None:
        return head
    tail=head
    cnt=1

    while tail.next is not None:
        cnt+=1
        tail=tail.next

    
    if k%cnt==0:
        return head

    k=k%cnt
    
    tail.next=head

    pos=cnt-k-1
    temp=head

    for _ in range(pos):
        temp=temp.next

    head=temp.next
    temp.next=None

    return head
```
