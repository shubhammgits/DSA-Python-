## Length of loop

```python
def sol(head):
    temp=head
    travel=0
    mydict = {}

    while temp is not None:

        if temp in mydict:
            return travel - mydict[temp] 
           
        mydict[temp] = travel
        travel+=1
        temp=temp.next

    return 0
            
```

```python
#optimal:

def sol(head):
    slow=head
    fast=head
    cnt=0

    while fast is not None and fast.next is not None:
        slow=slow.next
        fast=fast.next.next

        if slow==fast:
            slow=slow.next
            cnt+=1

            while slow!=fast:
                slow=slow.next
                cnt+=1

            return cnt

    return 0
            
```
