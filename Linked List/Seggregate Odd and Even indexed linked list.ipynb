## Seggregate Odd and Even indexed linked list

```python
#brute:

def sol(head):
    temp=head
    mylist=[]

    while temp is not None:
        mylist.append(temp.val)
        temp=temp.next.next

    temp=head.next

    while temp is not None:
        mylist.append(temp.val)
        temp=temp.next.next

    temp=head
    for i in range(len(mylist)):
        temp.val = mylist[i]
        temp=temp.next

    return head
```

```python
#optimal: 2 pointers

def sol(head):
    odd=head
    even=head.next

    evenHead = even

    while odd is not None and odd.next is not None:
        odd.next=odd.next.next
        odd=odd.next.next

        even.next=even.next.next
        even=even.next.next

    odd.next=evenHead

    return head

```
