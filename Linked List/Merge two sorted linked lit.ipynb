## Merge two sorted linked list

```python
#brute:

def sol(head1,head2):
    arr=[]
    temp1=head1
    temp2=head2


    while temp1 is not None:
        arr.append(temp1.val)
        temp1=temp1.next

    while temp2 is not None:
        arr.append(temp2.val)
        temp2=temp2.next

    arr.sort()

    head = convertFunc(arr)

    return head


def convertFunc(arr):

    dummy = ListNode(-1)
    curr = dummy

    for num in arr:
        curr.next=ListNode(num)
        curr=curr.next

    return dummy.next
```
