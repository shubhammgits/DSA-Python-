```python
#middle:

#brute:

def sol(head):
    temp = head
    cnt=0

    while temp is not None:
        cnt+=1
        temp = temp.next

    temp=head
    for i in range(cnt//2):
        temp=temp.next

    return temp
```

```python
#optimal: tortoise-hare approach

def sol(head):
    slow=head
    fast=head

    while fast is not None and fast.next is not None:
        slow=slow.next
        fast=fast.next.next

    return slow
```

```python
#reverse a linked list:

#brute: using stack

def sol(head):
    temp=head
    st = []

    while temp is not None:
        st.append(temp.val)
        temp=temp.next

    temp=head

    while temp is not None:
        newVal = st.pop()
        temp.val = newVal
        temp = temp.next

    return head
```
