## nodes creation

```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None

node1 = Node(5)
node2 = Node(10)
node3 = Node(7)
node4 = Node(9)

node1.next = node2
node2.next = node3
node3.next = node4
```

## Create Singly Linked List class

```python
class Node:
    def __init__(self,val):
        self.val=val
        self.next=None

class singlyLL:
    def __init__(self):
        self.head=None

    def append(self,val):
        new_node = Node(val)

        if self.head == None:
            self.head = new_node

        else:
            curr = self.head
            while curr.next != None:
                curr = curr.next

            curr.next = new_node


    
    def traversal(self):

        if self.head == None:
            print('SLL is empty')
        
        else:
            curr = self.head
            while curr.next != None:
                print(curr.val, end=' ')
                curr = curr.next
            print()


sll = singlyLL()
```

```python
sll.traversal()
```
