## Create Singly Linked List class

```python
class Node:
    def __init__(self,val):
        self.val=val
        self.next=None

class singlyLL:
    def __init__(self):
        self.head=None

    def append(self,val):
        new_node = Node(val)

        if self.head == None:
            self.head = new_node

        else:
            curr = self.head
            while curr.next != None:
                curr = curr.next

            curr.next = new_node


    
    def traversal(self):

        if self.head == None:
            print('SLL is empty')
        
        else:
            curr = self.head
            while curr != None:
                print(curr.val, end=' ')
                curr = curr.next
            print()


    

    def insert(self,val,position):
        new_node = Node(val)

        if position == 0:
            new_node.next = self.head
            self.head = new_node
            return

        current = self.head
        prev_node = None
        cnt = 0

        while current != None and cnt<position:
            prev_node = current
            current = current.next
            cnt+=1

        if cnt!=position:
            print('Invalid posiiton')
            return

        prev_node.next = new_node
        new_node.next = current




    def delete(self,val):
        
        if self.head is None:
            print('SLL is empty')
            return
        
        temp = self.head
        if temp.val == val:
            self.head = temp.next
            del temp
            return

        prev = None

        while temp is not None and temp.val != val:
            prev = temp
            temp = temp.next

        if temp is None:
            print('Not found')
            return

        prev.next = temp.next
        del temp

sll = singlyLL()
```
