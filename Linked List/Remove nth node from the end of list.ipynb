## Remove the nth node from end of the linked list

```python
#brute:

def sol(head,n):
    temp=head
    cnt=0

    while temp is not None:
        cnt+=1
        temp=temp.next

    if n==cnt:
        head=head.next
        return head

    pos=cnt-n
    temp=head
    prev=None

    for i in range(pos):
        prev=temp
        temp=temp.next

    prev.next=temp.next
    del temp

    return head

    
```
