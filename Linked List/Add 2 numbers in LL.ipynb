## Add two numbers in linked list

```python
#using dummy node

def sol(head1,head2):
    t1=head1
    t2=head2

    dummy=ListNode(-1)
    curr=dummy
    carry=0

    while t1 is not None or t2 is not None:
        currSum = carry

        if t1:
            currSum=currSum+t1.val
        
        if t2:
            currSum=currSum+t2.val

        carry = currSum//10
        newNode = ListNode(currSum%10)

        curr.next = newNode
        curr=curr.next

        if t1:
            t1=t1.next
        if t2:
            t2=t2.next

    if carry:
        newNode = ListNode(carry)
        curr.next = newNode

    
    return dummy.next
```
