## Sort Linked List of 0's, 1's and 2's

```python
#brute:

def sol(head):
    cnt0=0
    cnt1=0
    cnt2=0

    temp=head

    while temp is not None:
        if temp.val == 0:
            cnt0+=1

        elif temp.val==1:
            cnt1+=1

        else:
            cnt2+=1

        temp=temp.next

    
    temp=head

    while cnt0>0:
        temp.val=0
        cnt0-=1
        temp=temp.next

    while cnt1>0:
        temp.val=1
        cnt1-=1
        temp=temp.next

    while cnt2>0:
        temp.val=2
        cnt2-=1
        temp=temp.next

    return head
```
