## Merge k sorted lists

```python
#brute:

def sol(lists):
    k=len(lists)
    arr=[]

    for i in range(k):

        temp = lists[i]

        while temp is not None:
            arr.append(temp.val)
            temp=temp.next

    arr.sort()

    head = convert(arr)

    return head


def convert(arr):

    dummy=ListNode(-1)
    temp=dummy

    for num in arr:
        curr = ListNode(num)
        temp.next = curr
        temp=temp.next

    return dummy.next
```
