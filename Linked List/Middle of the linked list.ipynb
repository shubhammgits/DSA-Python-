## Middle of the linked list

```python
def sol(head):
    temp=head
    cnt=0
    while temp is not None:
        cnt+=1
        temp=temp.next

    temp=head

    for i in range(0,cnt//2):
        temp=temp.next

    return temp
```

```python
#optimal: tortoise-hare approach

def sol(head):
    slow=head
    fast=head

    while fast is not None and fast.next is not None:
        slow=slow.next
        fast=fast.next.next

    return slow
```
