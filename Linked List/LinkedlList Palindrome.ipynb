## LinkedList Palindrome

```python
def sol(head):

    if head is None or head.next is None:
        return True

    # Find the middle
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

    # Skip the middle node for odd-length lists
    if fast is not None:
        slow = slow.next

    # Reverse the second half
    prev = None
    curr = slow

    while curr is not None:
        front = curr.next
        curr.next = prev
        prev = curr
        curr = front

    # Compare both halves
    first = head
    second = prev

    while second is not None:
        if first.val != second.val:
            return False

        first = first.next
        second = second.next

    return True
```
