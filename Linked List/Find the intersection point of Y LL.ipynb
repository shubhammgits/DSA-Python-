## Find the intersection point of Y LL

```python
#brute: using hashing

def sol(head1, head2):
    temp1=head1
    temp2=head2

    freq={}

    while temp1 is not None:
        freq[temp1]=1
        temp1=temp1.next

    while temp2 is not None:

        if temp2 in freq:
            return temp2
        temp2=temp2.next

    
    return None
```

```python
#bettre: length based appraoch

def sol(head1, head2):
    t1=head1
    t2=head2

    l1=0
    l2=0

    while t1 is not None:
        l1+=1
        t1=t1.next

    while t2 is not None:
        l2+=1
        t2=t2.next


    if l1<l2:
        return collisionPoint(head1,head2,l2-l1)
    else:
        return collisionPoint(head2,head1,l1-l2)


def collisionPoint(smallerHead, largerHead, diff):

    temp1=smallerHead
    temp2=largerHead

    for i in range(diff):
        temp2=temp2.next

    while temp1!=temp2:
        temp1=temp1.next
        temp2=temp2.next

    return temp1
```

```python
## optimal : 2 pointers

def sol(head1,head2):

    if head1==None or head2==None:
        return None

    temp1=head1
    temp2=head2

    while temp1!=temp2:

        temp1=temp1.next
        temp2=temp2.next

        if temp1==temp2:
            return temp1

        if temp1==None:
            temp1=head2
        if temp2==None:
            temp2=head1

    return temp1
```
