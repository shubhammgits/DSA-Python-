```python
def isPalindrome(num):
    copy=num
    rev=0

    while copy>0:
        lastD=copy%10
        rev=rev*10+lastD
        copy=copy//10

    if rev==num:
        return f'{num} is a palindrome'
    else:
        return f'{num} is not a palindrome'
```

```python
isPalindrome(111)
```

```python
#second largest element:

def secondL(nums):
    largest=nums[0]
    for i in range(1,len(nums)):
        if nums[i]>largest:
            largest=nums[i]

    secondl=-1
    for i in range(len(nums)):
        if nums[i]>secondl and nums[i]!=largest:
            secondl=nums[i]

    return secondl
```

```python
nums=[1,2,4,7,7,5]
secondL(nums)
```

```python
#second smallest:

def secondS(nums):
    smallest=nums[0]
    for i in range(1,len(nums)):
        if nums[i]<smallest:
            smallest=nums[i]

    seconds=float('inf')
    for i in range(len(nums)):
        if nums[i]<seconds and nums[i]>smallest:
            seconds=nums[i]

    return seconds
```

```python
#optimal:

def reverse(arr,i,j):
    while i<j:
        arr[i],arr[j]=arr[j],arr[i]
        i+=1
        j-=1

def leftRotate(arr,k):
    n=len(arr)
    k=k%n

    reverse(arr,0,k-1)
    reverse(arr,k,n-1)
    reverse(arr,0,n-1)

    return arr
```

```python
arr = [1,2,3,4,5,6,7]
leftRotate(arr,3)
```

```python
#optimal:

arr=[1,0,2,3,4,0,0,4,5,1]
n=len(arr)

def moveZeroes(arr,n):
    j=-1
    for i in range(n):
        if arr[i]==0:
            j=i
            break

    if j==-1:
        return arr

    for i in range(j+1,n):
        if arr[i]!=0:
            arr[j],arr[i]=arr[i],0
            j+=1

    return arr
        
```

```python
#optimal : 2pointer if array is sorted:

def sol(arr,n,target):
    left=0
    right=n-1

    arr.sort()

    while left<right:
        if arr[left]+arr[right]==target:
            return arr[left],arr[right]

        elif arr[left]+arr[right]<target:
            left+=1

        else:
            right-=1

arr = [2,6,5,8,11]
sol(arr,len(arr),target=14)
```

```python
#optimal:

arr = [1,0,0,2,1,1,1,0,0,2,2,1,0,1,2]
n = len(arr)

def sol(arr,n):
    low=0
    mid=0
    high=n-1

    while mid<=high:
        if arr[mid]==0:
            arr[low],arr[mid]=arr[mid],arr[low]
            low+=1
            mid+=1

        elif arr[mid]==1:
            mid+=1

        else:
            arr[mid],arr[high]=arr[high],arr[mid]
            high-=1

    return arr

sol(arr,n)
```

```python
#optimal: moores voting algo

arr = [2,2,3,3,1,2,2]
n=len(arr)

def sol(arr,n):
    elem=0
    cnt=0
    
    for i in range(n):
        if cnt==0:
            elem=arr[i]
            cnt=1

        elif arr[i]==elem:
            cnt+=1

        else:
            cnt-=1

    verifyCnt=0
    for i in range(n):
        if arr[i]==elem:
            verifyCnt+=1

    if verifyCnt>=n/2:
        return elem

    else:
        return -1

sol(arr,n)
```

```python
arr = [3,1,-2,-5,2,-4]
n = len(arr)

#optimal

def sol(arr,n):
    posInd=0
    negInd=1

    ans = [0]*n

    for i in range(n):
        if arr[i]>0:
            ans[posInd]=arr[i]
            posInd+=2
        else:
            ans[negInd]=arr[i]
            negInd+=2

    return arr

sol(arr,n)
```

```python
#alternate numbers

arr = [1,2,-4,-5,3,6,8,9,0]
n = len(arr)

pos=[]
neg=[]

for i in range(n):
    if arr[i]<0:
        neg.append(arr[i])
    else:
        pos.append(arr[i])

miniAlt = min(len(pos),len(neg))


for i in range(miniAlt):
    arr[i*2]=pos[i]
    arr[2*i+1]=neg[i]


arr
```

```python
#longest coonsecutive sequence:

arr = [102,4,100,1,101,3,2,1,1]
n = len(arr)
longest = -1

myset = set()
for num in arr:
    myset.add(num)

for num in myset:
    if num-1 not in myset:
        current=num
        cnt=1

        while current+1 in myset:
            current+=1
            cnt+=1

        longest=max(longest,cnt)

longest
```
